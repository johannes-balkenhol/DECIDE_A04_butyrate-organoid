## Monotonic responders (same sign at 5 mM and 10 mM, sig at 10 mM)

```python
def monotonic(d5, d10):
    j = d5[['log2FoldChange']].join(d10[['log2FoldChange','padj','symbol']], lsuffix='_5', rsuffix='_10')
    j = j.dropna()
    same = np.sign(j['log2FoldChange_5'])==np.sign(j['log2FoldChange_10'])
    return j[same & (j['padj']<0.05) & (j['log2FoldChange_10'].abs()>1)]
mono_mock = monotonic(m5,m10); mono_cd = monotonic(c5,c10)
print('monotonic butyrate genes — Mock:', len(mono_mock), ' Cd:', len(mono_cd))
mono_cd.sort_values('log2FoldChange_10', ascending=False)[['symbol','log2FoldChange_5','log2FoldChange_10','padj']].head(15)
```
